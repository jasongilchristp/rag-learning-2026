# JSON Loader

```python
from langchain_classic.document_loaders.json_loader import JSONLoader
from pathlib import Path
from pprint import pp
```

```python
# define path to JSON file

file_path = Path("../data/apparels.json")

print(file_path.exists())
```

```python
def metadata_func(record: dict, metadata: dict) -> dict:
    metadata["product_name"] = record["productName"]
    metadata["category"] = record["category"]
    metadata["price"] = record["price"]

    # del unwanted columns
    del metadata["seq_num"]
    return metadata
```

```python
from importlib.readers import FileReader
# create JSON loader

loader = JSONLoader(
    file_path = file_path,
    jq_schema = ".products[]",
    content_key = "Description",
    metadata_func = metadata_func,
)
```

```python
# load the documents 
documents = loader.load()
```

```python
for doc in documents:
    print(doc.page_content, end="\n\n")
```
