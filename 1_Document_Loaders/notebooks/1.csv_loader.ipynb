# CSV LOADER

```python
from pathlib import Path
import pandas as pd
from langchain_community.document_loaders import CSVLoader, DataFrameLoader

# Define path to CSV file relative to notebook directory
file_path = Path("../data/organizations.csv")
```

## 1. Using LangChain's CSVLoader

```python
# Load using CSVLoader directly from file path
loader = CSVLoader(file_path=str(file_path), source_column="Name")
docs = loader.load()

print(f"Loaded {len(docs)} documents using CSVLoader\n")
print("Sample Document 0 Content:")
print(docs[0].page_content)
print("\nSample Document 0 Metadata:")
print(docs[0].metadata)
```

## 2. Using DataFrameLoader with Pandas

```python
# Read CSV using pandas
df = pd.read_csv(file_path, encoding="utf-8")

# DataFrameLoader converts rows of a Pandas DataFrame into Document objects
# page_content_column is the column containing text for the Document content
loader = DataFrameLoader(
    df,
    page_content_column="Description",
    metadata_columns=["Name", "Website", "Country", "Founded", "Industry", "Number of employees"]
)
docs = loader.load()

print(f"Loaded {len(docs)} documents using DataFrameLoader\n")
print("Sample Document 0 Content:")
print(docs[0].page_content)
print("\nSample Document 0 Metadata:")
print(docs[0].metadata)
```
